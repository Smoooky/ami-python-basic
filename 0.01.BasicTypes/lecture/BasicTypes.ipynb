### Арифметические операции

Сложение, вычитание, умножение и модуль числа:

````python
a = 5
b = 10

a + b, a - b, a * b, abs(-a)
````

Деление представлено тремя операциями, и различать их важно:

* `/` — обычное деление, результат всегда `float`;
* `//` — целочисленное деление (отбрасывание дробной части вниз);
* `%` — остаток от деления.

````python
a = 5
b = 10

print(b / a, type(b / a))
print(b // a, type(b // a))
print(a % b, type(a % b))
````

`//` и `%` определены и для `float`:

````python
a = 5
b = 1.12

a // b, a % b
````

Смысл этих операций — деление с остатком. Для $7$ и $3$:

$$7 = 3 \cdot 2 + 1$$

где $2$ — неполное частное, $1$ — остаток:

````python
7 // 3, 7 % 3
````

Знак остатка в Python совпадает со знаком **делителя**, а не делимого:

````python
-15 % 6, 15 % -6
````

Возведение в степень — оператор `**`, показатель может быть вещественным:

````python
a = 5
b = 2.2

a**b
````

Для операций вида «изменить переменную и записать результат в неё же»
существует составное присваивание: `a += b` эквивалентно `a = a + b`.

````python
a = 5
b = 2

a += b
print(a)

a -= b
print(a)

a *= b
print(a)

a //= b
print(a)

a **= b
print(a)
````

## Строки

Строка (`str`) — **неизменяемая последовательность символов**. Записывается в
одинарных или двойных кавычках; для Python это одно и то же, важно лишь чтобы
открывающая и закрывающая совпадали.

````python
s = "4594520"
type(s)
````

### Индексация и срезы

Нумерация начинается с нуля. Отрицательные индексы отсчитываются с конца:
`-1` — последний символ.

````python
s = "4594520"

s[0], s[-1]
````

Общая форма среза:

$$s[i:j:k]$$

«взять элементы с `i`-го включительно по `j`-й не включительно с шагом `k`».
Любую из трёх частей можно опустить.

````python
s = "4594520"

s[:3], s[2:5], s[::2], s[::-1]
````

Индексация и срезы работают одинаково для всех последовательностей — строк,
списков, кортежей. Ниже они встретятся снова.

### Операции над строками

Сложение строк даёт конкатенацию, умножение на целое число — повторение:

````python
s = "4594520"

s + "123", s * 2
````

Неизменяемость означает, что отдельный символ заменить нельзя — попытка
приводит к ошибке:

````python
s = "4594520"
s[0] = "1"
````

### Методы строк

Полный список — в [документации](https://docs.python.org/3/library/stdtypes.html#string-methods).
Наиболее употребимые:

* `len(s)` — длина строки;
* `s.count(ss)` — число непересекающихся вхождений подстроки `ss`;
* `s.find(ss)` — индекс **первого** вхождения `ss`, или `-1`, если вхождений нет;
* `s.rfind(ss)` — индекс **последнего** вхождения, или `-1`;
* `s.isdigit()` — состоит ли строка только из цифр (проверка перед приведением к числу);
* `s.islower()`, `s.isupper()` — все ли буквы строчные или заглавные;
* `s.strip(chars)` — убрать с **обоих концов** строки любые символы, входящие в `chars`; без аргумента убирает пробельные символы;
* `s.replace(old, new)` — заменить все вхождения `old` на `new`;
* `s.split(sep)` — разбить строку по разделителю `sep`; без аргумента — по пробельным символам;
* `sep.join(strings)` — соединить последовательность строк через разделитель `sep`.

````python
s = "I love python (maybe)"
s_1 = "1234"

(
    len(s),
    s.count("o"),
    s.find("o"),
    s.rfind("o"),
    s.find("-"),
    s.isdigit(),
    s_1.isdigit(),
    s.islower(),
)
````

Обратите внимание на `strip`: он работает как набор символов, а не как
подстрока, и удаляет только с краёв.

````python
"aXbXa".strip("aX"), "  текст  ".strip(), "abcba".replace("b", "")
````

````python
full_s = " ".join(["I love python", "1234"])

print(full_s)
print(full_s.split())
````

## Байтовые типы

Строка состоит из символов, а файлы и сеть оперируют байтами. Перевод между
ними называется кодированием. Тип `bytes` — неизменяемая последовательность
байт, записывается с префиксом `b`:

````python
data = b"hello"

data, type(data)
````

Метод `str.encode()` кодирует строку в байты, `bytes.decode()` выполняет
обратное преобразование. По умолчанию используется кодировка UTF-8.

````python
s = "café"
b = s.encode()

b, type(b), len(b)
````

````python
b.decode(), b.decode("utf8")
````

Длина строки — 4 символа, длина её байтового представления — 5 **байт**:
символ `é` в UTF-8 кодируется двумя байтами. Символ и байт — не одно и то же,
и для не-ASCII текста это различие существенно.

````python
len("café"), len("café".encode())
````

`bytearray` устроен так же, но изменяем:

````python
mutable_data = bytearray(b"abc")
mutable_data[0] += 1

mutable_data
````

## Сравнения и логические операторы

### Операторы сравнения

Результат сравнения — значение типа `bool`.

````python
a = 1
b = 2
c = 1

print(a > b)  # больше
print(a >= c)  # больше или равно
print(a < b)  # меньше
print(a <= c)  # меньше или равно
print(a == c)  # равно; обратите внимание на два знака
print(a != b)  # не равно
````

Значения разных типов сравнивать на равенство можно: `==` не вызывает ошибку, а
возвращает `False`, если объекты неравны. Числа разных числовых типов при этом
сравниваются по значению.

````python
1 == "1", 1 == 1.0, 1 != "1"
````

А вот упорядочивающие сравнения (`<`, `>`, `<=`, `>=`) для несопоставимых типов
завершаются ошибкой:

````python
1 < "1"
````

### Логические операторы

* `and` — логическое И;
* `or` — логическое ИЛИ;
* `not` — логическое НЕ.

````python
a = 1
b = 2
c = 1

print(a == c and b > a)
print(a != c or a != b)
print(not a)
````

Сравнения можно объединять в цепочки — запись читается как в математике:

````python
a = 1

-1 <= a <= 2
````

### Оператор `in`

Проверяет принадлежность элемента последовательности или множеству:

````python
1 in [1, 2, 3], "lo" in "hello", 5 in {1, 2, 3}
````

### `==` и `is`

`==` сравнивает **значения**, `is` — является ли это одним и тем же объектом.
Для проверки на `None` принято использовать `is`, поскольку `None` существует в
единственном экземпляре. Подробнее об объектах и ссылках — в лекции про
изменяемость.

````python
a = None

a is None, a == None
````

### Короткое замыкание

Логические операторы вычисляют правую часть только тогда, когда результат ещё
не определён левой. Для `and`: если слева `False`, правая часть не вычисляется.
Для `or`: если слева `True`, правая часть не вычисляется.

Практическое следствие: правую часть можно защитить проверкой в левой. В
примере ниже деления на ноль не происходит — до него дело не доходит.

````python
x = 0

x != 0 and 10 / x
````

````python
x = 0

x == 0 or 10 / x
````

Если убрать защиту, выражение завершится ошибкой:

````python
x = 0

10 / x
````

Это не оптимизация интерпретатора, а гарантия языка, на которую можно
опираться.

### Сравнение последовательностей

Последовательности сравниваются лексикографически: поэлементно слева направо,
а при совпадении общего начала короче считается меньшей.

````python
print("apple" < "banana")
print("abc" < "abcd")

print([1, 2, 3] < [1, 2, 4])
print([1, 2] < [1, 2, 3])

print((1, 2) < (1, 3))
````

## Битовые операции

Операции над двоичным представлением целых чисел: `&` — И, `|` — ИЛИ,
`^` — исключающее ИЛИ, `~` — инверсия, `<<` и `>>` — сдвиги.

````python
bin(10), bin(13)
````

````python
10 & 13, 10 | 13, 10 ^ 13
````

````python
~1, bin(1)
````

`1 << 2 + 3` — это `1 << 5`, а не `(1 << 2) + 3`. В неочевидных случаях
расставляйте скобки.

````python
1 << 2 + 3, (1 << 2) + 3
````

## Разбор строки с числами

Числа редко приходят по одному: чаще это строка, в которой они записаны через
пробел — строка из файла, аргумент функции, ответ сервиса. Чтобы считать
сумму или максимум, такую строку сначала превращают в список чисел.

Разбор состоит из трёх шагов:

1. `str.split()` разбивает строку на список строк по пробельным символам;
   без аргумента подряд идущие пробелы считаются одним разделителем, а пробелы
   по краям отбрасываются;
2. `map(int, ...)` применяет `int` к каждому элементу;
3. `list(...)` собирает результат в список.

````python
line = "10 20 30"

line.split()
````

````python
"  7   8  ".split()
````

После `split` элементы — по-прежнему строки. Их нужно привести к числам:

````python
line = "10 20 30"
numbers = list(map(int, line.split()))

numbers, type(numbers[0])
````

`list` здесь обязателен: сам по себе `map` списка не даёт, он лишь описывает
преобразование.

````python
map(int, "10 20 30".split())
````

Если количество чисел известно заранее, результат удобно сразу разложить по
переменным — слева перечисляются имена через запятую:

````python
a, b = map(int, "3 4".split())

a, b
````

К готовому списку чисел применимы встроенные функции: `len` — количество,
`sum` — сумма, `max` и `min` — наибольшее и наименьшее.

````python
numbers = list(map(int, "10 20 30".split()))

len(numbers), sum(numbers), max(numbers), min(numbers)
````

### Объявление

Функция объявляется словом `def`, за ним имя, круглые скобки со списком
параметров и двоеточие. Тело записывается с отступом — так же, как и в других
конструкциях языка.

```
def <имя>(<параметры>):
    <тело>
    return <результат>
```

````python
def double(x):
    return x * 2
````

Объявление само по себе ничего не вычисляет. Тело выполняется только при
**вызове** — когда имя функции написано со скобками:

````python
double(21)
````

`x` в объявлении — параметр, имя для значения, которое придёт извне. При вызове
`double(21)` параметр `x` получает значение `21`. Имя параметра существует

### return

`return` задаёт результат и **немедленно завершает** функцию: строки после него
не выполняются.

````python
def sign_text(number):
    return "положительное"
    print("эта строка недостижима")


sign_text(5)
````

Вернуть можно несколько значений сразу — через запятую. Получится знакомый
кортеж, и его удобно распаковывать в отдельные переменные.

````python
def divide(a, b):
    return a // b, a % b


divide(17, 5)
````

````python
quotient, remainder = divide(17, 5)

quotient, remainder
````

### `return` — это не `print`

Частая ошибка новичков. `print` показывает значение человеку в стандартном
выводе, а `return` отдаёт его программе. Функция без `return` возвращает `None`,
даже если что-то напечатала.

````python
def printing(x):
    print(x * 2)


def returning(x):
    return x * 2


result_1 = printing(21)
result_2 = returning(21)

result_1, result_2
````

Видно, что `printing` напечатала `42`, но вернула `None` — использовать это
значение дальше нельзя. **Тесты в домашнем задании проверяют именно
возвращаемое значение**, поэтому в задачах нужен `return`, а не `print`.

### Аннотации типов

В задачах курса сигнатуры записаны с аннотациями:

```python
def seconds_to_time(total: int) -> tuple[int, int, int]: ...
```

Читается так: параметр `total` предполагается целым числом, а функция
возвращает кортеж из трёх целых. Аннотации — это пояснение для читателя и для
инструментов вроде `mypy`. Сам Python их **не проверяет** и не приводит
значения к указанным типам:

````python
def double_annotated(x: int) -> int:
    return x * 2


double_annotated("ab")  # аннотация говорит int, но строка тоже сработает
````

Менять аннотации в задачах не нужно — сигнатура уже написана за вас.

### Как устроена задача

В директории задачи лежит файл с готовой сигнатурой и заглушкой в теле:

```python
def greet(name: str) -> str:
    raise NotImplementedError("Implement me")
```

Строку с `raise` нужно заменить своим кодом, сохранив имя функции и список
параметров без изменений. Вызывать функцию самому не требуется — это делают
тесты, которые лежат рядом в `test_public.py`.
