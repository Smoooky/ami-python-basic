## Множества

Скорее всего вы знаете, что такое множества, исходя из математики. Множество - это набор уникальных элементов.

Казалось бы, ту же роль мог играть список. Разница в скорости поиска: у списка он требует перебора элементов, у множества — вычисления хеша. Сравним на миллионе проверок:

```python
a = [i for i in range(100000)]
b = set(a)
```

```python
%%timeit

1000000 in a
```

```python
%%timeit

1000000 in b
```

Видим, что поиск в множестве намного быстрее чем в списке. Как это работает? На самом деле множество - хеш-таблица. Саму хеш-таблицу вы разберете на курсе алгоритмов и структур данных, а вот что такое хеш и хешируемость мы обсудим на нашем курсе.

### Операции

Методы множеств:

* `s.add(elem)` - добавить элемент во множество (если элемент уже есть, то ничего не изменится)

* `clear()` - очистить множество

* `copy()` - скопировать множество

* `s.discard(elem)` / `s.remove(elem)` / `s.pop()` - разные методы удаления (первое - не ругнется, если попробовать убрать элемент не из множества, второй - ругнется, третий - просто вытаскивает рандомный элемент и возвращает его)

* `difference` / `difference_update()` / `-` - разность

* `union()` / `|` - объединение множеств

* `intersection()` / `&` - пересечение множеств

* `issubset()` / `isdisjoint()` / `issuperset()` - проверка на подмножество, наличие пересечений и проверка на супермножество (один находится в другом)

* `symmetric_difference` / `^` - симметричная разность

* `len(s)` - узнать число элементов во множестве

* `==`, `<=`, `>=` - проверки на равенство (поэлементно), является ли одно множество под(над)множеством другого

```python
a = {1, 2, 3, 5}
b = {4, 5, 6}
```

```python
a.union(b), a | b
```

```python
a.intersection(b), a & b
```

```python
a.difference(b), a - b
```

```python
a.symmetric_difference(b), a ^ b
```

Сравнение множеств:

```python
a = {1, 2}
b = {1, 2, 3}

# Подмножество
a.issubset(b)
assert a <= b

# Собственное подмножество
assert a < b

# Надмножество
b.issuperset(a)
assert b >= a

# Собственное надмножество
assert b > a
```

Прежде чем идти дальше — про `sorted`, он понадобится ниже. Функция принимает

отсортировать пары.

```python
sorted([(2, "b"), (1, "z"), (1, "a")])
```

Обратный порядок задаётся аргументом `reverse`:

```python
sorted([3, 1, 2], reverse=True)
```

```python
# Эффективный способ удаления дубликатов
items = [1, 2, 3, 2, 1, 4, 5]
unique = set(items)
assert unique == {1, 2, 3, 4, 5}

# Порядок элементов множества не определён, поэтому сравнивать список с
# конкретным порядком нельзя. Если нужен предсказуемый порядок — сортируем:
assert sorted(unique) == [1, 2, 3, 4, 5]
```

```python
# Быстрая фильтрация с использованием множеств
all_items = [1, 2, 3, 4, 5, 6, 7, 8, 9]
valid_items = {2, 4, 6, 8}

filtered = [item for item in all_items if item in valid_items]
assert filtered == [2, 4, 6, 8]
```

```python
# Эффективный поиск общих элементов в нескольких коллекциях
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
list3 = [5, 6, 7, 8, 9]

common = set(list1) & set(list2) & set(list3)
assert common == {5}
```

```python
# Пустое множество ложно в булевом контексте
s = set()
if not s:
    print("Множество пустое")

# Непустое множество истинно
s = {1, 2, 3}
if s:
    print("Множество не пустое")
```

### Set Comprehensions

Множество, как и список, можно наполнять в цикле. Например, отобрать слова
длиннее пяти букв — дубликаты при этом схлопнутся сами:

```python
words = ["ананас", "банан", "вишня", "банан"]

s = set()
for word in words:
    if len(word) > 5:
        s.add(word)
s
```

Однако, в питоне принято конструировать объекты с использованием comprehensions:

```python
sd = {word for word in words if len(word) > 5}
sd
```

```python
squares = {x**2 for x in range(10)}
assert squares == {0, 1, 4, 9, 16, 25, 36, 49, 64, 81}

even_squares = {x**2 for x in range(10) if x % 2 == 0}
assert even_squares == {0, 4, 16, 36, 64}
```

Такая запись короче и не требует заводить пустой объект перед циклом — в Python она считается предпочтительной.

### Хешируемость

```python
a = {1}
a.add([15, 20])
```

Тут мы сталкиваемся с той проблемой, что есть **хешируемые** и **нехешируемые** объекты.

Для начала давайте поймем что такое хеш-функция.

Пусть у нас есть $U$ — универсум всех возможных входных данных (ключей), $S = \{0, 1, 2, \ldots, m-1\}$ — множество целочисленных хеш-значений размера $m$.

Хеш-функция $h$ определяется как отображение:

$h: U \rightarrow S$

Хешируемый объект - это:
- Объект может быть передан аргументом в функцию $h$ (имеет магический метод `__hash__`), то есть можно получить **хеш** от объекта.
- Сравним с другими объектами: имеет магический метод `__eq__` (об этом мы еще отдельно поговорим).
- $\forall x \in U, h(x) = \text{const}$
- $\forall x, y \in U, x = y \Rightarrow h(x) = h(y)$

```python
hash(1), hash(2), hash(10)
```

Вопрос: что будет выведено в данном случае?

```python
hash(-1), hash(-2)
```

Хеши совпадают: `hash(-1)` равен `-2`. Значение `-1` зарезервировано в CPython
под признак ошибки — функция `__hash__` возвращает его, когда вычисление хеша
не удалось. Чтобы настоящий хеш никогда не совпал с признаком ошибки, для `-1`
результат подменяется на `-2`. Это деталь реализации CPython, а не требование
языка.

```python
{-1, -2}
```

```python
items = ["apple", 42, (1, 2), frozenset([3, 4])]
for item in items:
    print(f"{item!r}: hash = {hash(item)}")
```

От кортежа можно получить хеш, соответственно он может быть ключом в множестве:

```python
t = (1, 2, (3, 4))
hash(t)
```

Вопрос: что произойдет в данном примере?

```python
t = (1, 2, [3, 4])
hash(t)
```

Почему так? Потому что список - изменяемый объект внутри кортежа.

Важно понимать, что хеш-функция объекта с вложением - некоторое сочетание результатов хеш-функции вложенных объектов. То есть, чтобы получить хеш от кортежа нужно получить хеш от всего, что в него вложено. Тут то и возникает проблема с тем, что мы не можем получить хеш у списка, так как это изменяемый объект, или, корректнее говоря, у списка отсутствует метод `__hash__`.

Честно говоря я не раздобыл какая действительно функция лежит под капотом, но пример функции ниже выглядит вполне приемлемым:

Для кортежа $T = (x_1, x_2, \ldots, x_n)$:
$$
h_{\text{tuple}}(T) = \left( c_{\text{tuple}} + \sum_{i=1}^{n} h(x_i) \cdot p_i^{n-i} \right) \mod m
$$
где $c_{\text{tuple}}$ — константа для типа "кортеж", $p_i$ — простые числа, $m$ - размер словаря.

```python
a = set()
a.add((15, 20))
a
```

```python
a.add(frozenset({15, 20}))
a
```

А вот если положить список в замороженное множество, то можем спокойно брать хеш.

```python
t = (1, 2, frozenset([3, 4]))
hash(t)
```

### Оптимизации

Посмотрим, во что превращаются разные способы создания множества. Модуль `dis`
показывает байткод — инструкции, которые интерпретатор выполняет на самом деле.

```python
from dis import dis

dis("{1, 2}")
```

Литерал собирается одной инструкцией: константный кортеж превращается во
множество. Вызов `set([1, 2])` вместо этого требует загрузить имя `set`,
построить список и выполнить вызов — три лишних действия на каждом исполнении.
Поэтому непустое множество дешевле записывать литералом.

Важная оговорка: к **пустому** множеству это не относится. Литерала пустого
множества в языке нет — `{}` создаёт пустой словарь, и единственный способ
получить пустое множество это `set()`.

## Массивы array

Список хранит **ссылки** на объекты произвольных типов, поэтому каждый элемент
это отдельный объект в памяти. Тип `array` из стандартной библиотеки хранит
значения одного типа подряд, как в C: код типа задаётся первым аргументом
(`"l"` — целое, `"d"` — вещественное).

Отсюда два следствия: структура занимает заметно меньше памяти и не тратит
время на распаковку объектов при обходе. Плата — однородность: положить в
`array` значение другого типа нельзя.

```python
import array

arr = array.array("l", (1, 2, 3, 4))
arr.append(5)

arr, arr.typecode, len(arr)
```

```python
arr.append(3.5)  # тип фиксирован
```

Разница в памяти на миллионе элементов — заметная. `__sizeof__` возвращает
размер самого объекта в байтах, без учёта объектов, на которые он ссылается,
поэтому для списка настоящий расход ещё больше.

```python
import array

numbers = list(range(1_000_000))
packed = array.array("l", numbers)

numbers.__sizeof__(), packed.__sizeof__()
```

Чтение и запись `array` идут в двоичном виде, поэтому обмен такими данными
через файл во много раз быстрее, чем разбор текстовых чисел.

## Словари

### Создание

Словарь (`dict`) в Python — встроенная структура данных, которая хранит коллекцию элементов в виде пар ключ-значение (key-value). 

Ключи выступают уникальными идентификаторами для доступа к соответствующим значениям. 

Словари реализованы с использованием хэш-таблиц, что обеспечивает эффективность операций добавления, удаления и поиска элементов.

```python
a = dict(one=1, two=2, three=3)
b = {"one": 1, "two": 2, "three": 3}
c = dict(zip(["one", "two", "three"], [1, 2, 3]))
d = dict([("two", 2), ("one", 1), ("three", 3)])
e = dict({"three": 3, "one": 1, "two": 2})
a == b == c == d == e
```

Рассмотрим подробнее создание с использованием функции zip:

```python
c = dict(zip(["one", "two", "three"], [1, 2, 3]))
c
```

zip попарно соединяет два итерируемых объекта (про это будем говорить на другой лекции).

```python
type(zip(["one", "two", "three"], [1, 2, 3]))
```

```python
list(zip(["one", "two", "three"], [1, 2, 3])) == [("one", 1), ("two", 2), ("three", 3)]
```

### Хешируемые ключи

Еще раз про хешируемость, но теперь в словарях:

```python
d = {
    1: "int",
    "a": "str",
    (1, 2): "tuple",
    frozenset([1, 2]): "frozenset",
    None: None,
}
```

Попытка вставить нехешируемый ключ вызывает ошибку `TypeError`.

```python
d = {
    [1, 2]: "list",
    {1, 2}: "set",
    {"a": 1}: "dict",
}
```

**Самый надёжный способ проверить, хешируем ли объект, — вызвать от него
`hash`.** Если объект нехешируем, вызов завершится `TypeError`.

Вопрос: что будет внутри словаря?

```python
{True: True, 1: 1, 1.0: 1.0}
```

Получится `{True: 1.0}` — одна пара. Ключи в словаре различаются по хешу и
равенству, а `True == 1 == 1.0` и все три хеша равны единице. Поэтому это один
и тот же ключ: он записывается при первой вставке и больше не меняется, а
значение перезаписывается каждой следующей. Отсюда несимметричный результат —
ключ от первого литерала, значение от последнего.

### Dict Comprehensions

Словарь, как список и множество, можно собрать включением:

```python
capitals = [
    ("Russia", "Moscow"),
    ("USA", "Washington"),
    ("China", "Beijing"),
]

d = {country: capital.upper() for country, capital in capitals}  # это dict comprehensions
d
```

Способ менее изящный и нерекомендуемый:

```python
d = {}
for country, capital in capitals:
    d[country] = capital.upper()
d
```

Можно удобно фильтровать внутри структуры создания словаря:

```python
d = {country: capital.upper() for country, capital in capitals if len(capital) < 8}
d
```

Паттерн матчинг со словарями:

```python
food = {"категория": "мороженое", "вкус": "ванильное", "цена": "110р"}

match food:
    case {"категория": "мороженое", **details}:
        print(f"Характеристики мороженого: {details}")
```

### Представления словаря

`keys()`, `values()` и `items()` возвращают не списки, а объекты-представления:
`dict_keys`, `dict_values`, `dict_items`. Они не копируют данные, а смотрят на
живой словарь — похожий по идее `memoryview` разбирается в лекции про ссылки и
изменяемость.

```python
d = dict(a=1, b=2, c=3)
values = d.values()

values, type(values), len(values), list(values)
```

Индексироваться по представлению нельзя — это не последовательность:

```python
values[0]
```

Зато представление отражает изменения словаря: отдельно обновлять его не нужно.

```python
d["d"] = 4

values
```

Главная практическая польза — `dict_keys` и `dict_items` поддерживают операции
множеств. Это позволяет сравнивать словари по ключам без промежуточных списков,
причём реализовано это на C.

```python
d1 = dict(a=1, b=2, c=3, d=4)
d2 = dict(b=20, d=40, e=50)

d1.keys() & d2.keys(), d1.keys() - d2.keys()
```

### Порядок ключей

```python
DIAL_CODES = [
    (86, "China"),
    (91, "India"),
    (1, "United States"),
    (62, "Indonesia"),
    (55, "Brazil"),
    (92, "Pakistan"),
    (880, "Bangladesh"),
    (234, "Nigeria"),
    (7, "Russia"),
    (81, "Japan"),
]
```

Сперва сложим в словарь все как есть:

```python
d1 = dict(DIAL_CODES)
print("d1:", d1.keys())
```

Теперь попробуем посортировать по коду:

```python
d2 = dict(sorted(DIAL_CODES))
print(f"d2: {d2.keys()}")
```

Теперь попробуем посортировать по названию страны:

```python
# кортежи сравниваются лексикографически, поэтому переставляем поля местами
by_country = sorted((country, code) for code, country in DIAL_CODES)
d3 = dict((code, country) for country, code in by_country)
print("d3:", d3.keys())
```

```python
assert d1 == d2 == d3
```

При этом все эти словари считаются равными, но ключи в них лежат в различном порядке:

```python
print("Без упорядочивания:")
for k, v in d1.items():
    print(f"({k}, {v})", end=" ")

print("\nС сортировкой по кодам:")
for k, v in d2.items():
    print(f"({k}, {v})", end=" ")

print("\nС сортировкой по названиям стран:")
for k, v in d3.items():
    print(f"({k}, {v})", end=" ")
```

Какой вывод из этого можно сделать? Не стоит завязывать какую-либо логику на порядок ключей в словаре - по дефолту ключи в словаря упорядочены в порядке их добавления в словарь (в современных версиях питона).
