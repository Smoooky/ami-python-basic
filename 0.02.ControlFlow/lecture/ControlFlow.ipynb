## Форматирование строк

### Префиксы строковых литералов

Перед кавычками может стоять префикс, меняющий смысл литерала:

* `f` — форматированная строка: выражения в фигурных скобках вычисляются;
* `r` — «сырая» строка: экранирующие последовательности вроде `\n` не
  обрабатываются и остаются двумя символами;
* `b` — байтовая строка (разбиралась в лекции 1);
* `u` — строка Unicode; нужен был для совместимости с Python 2, сейчас
  бессмысленен.

Префиксы `r` и `f` сочетаются: `rf"..."`.

### Обычный вывод и его ограничения

`print` умеет печатать несколько значений, но управлять их взаимным
расположением неудобно:

````python
a = 5

print("your number is:", a, ", great")
````

Лишние пробелы возникают из-за разделителя по умолчанию. Собирать строку
конкатенацией тоже неудобно: числа приходится приводить к `str` вручную.

````python
a = 5

print("your number is: " + str(a) + ", great")
````

### f-строки

Внутри f-строки фигурные скобки содержат выражение, которое вычисляется и
подставляется в текст. Приведение к строке происходит автоматически.

````python
a = 5

print(f"your number is: {a}, great")
````

Внутри скобок допустимо любое выражение, а не только имя переменной:

````python
a = 5
b = 3

print(f"{a} + {b} = {a + b}")
print(f"половина от {a} это {a / 2}")
````

### Спецификация формата

После двоеточия внутри скобок записывается формат — как именно печатать значение.
Наиболее употребимое:

* `.4f` — вещественное число с четырьмя знаками после точки;
* `_` или `,` — разделитель групп разрядов;
* `<`, `^`, `>` с числом — выравнивание по левому краю, по центру, по правому
  в поле заданной ширины;
* `.5` для строки — обрезать до пяти символов;
* `b`, `o`, `x` — двоичная, восьмеричная, шестнадцатеричная запись.

````python
value = 15.000005

print(f"{value:.4f}")
print(f"{value:.1f}")
````

````python
number = 1231234.5678

print(f"{number:_.2f}")
````

````python
text = "Hi"

print(f"[{text:<8}]")
print(f"[{text:^8}]")
print(f"[{text:>8}]")
````

````python
long_text = "Hello World"

print(f"{long_text:.5}")
````

````python
n = 255

print(f"{n:b}", f"{n:o}", f"{n:x}")
````

Отдельно полезен суффикс `=`: он печатает и само выражение, и его значение.
Удобно при отладке.

````python
x = 10

print(f"{x=}")
print(f"{x * 2=}")
````

### Сырые строки

В сырой строке обратный слеш не начинает экранирующую последовательность:

````python
print("a\nb")
print(r"a\nb")
````

Подробный справочник по синтаксису f-строк: [fstring.help](https://fstring.help/).

## Условные операторы

### if / elif / else

Синтаксис:

```
if <условие>:
    <действия>
elif <другое условие>:
    <действия>
else:
    <действия>
```

Блок задаётся отступом — фигурных скобок в Python нет. Условия проверяются
сверху вниз, выполняется тело **первого** истинного; `elif` и `else`
необязательны.

````python
x = 4

if x % 2 == 0:
    print(f"Число {x} чётное")
else:
    print(f"Число {x} нечётное")
````

````python
x = 20

if x % 10 == 0:
    print(f"{x} делится на 10")
elif x % 5 == 0:
    print(f"{x} делится на 5")
else:
    print(f"{x} не делится ни на 5, ни на 10")
````

Порядок ветвей важен: для `x = 20` истинны оба первых условия, но выполнится
только первое. Если поменять их местами, ветка с делением на 10 станет
недостижимой.

### Условное выражение

Если по условию выбирается одно из двух **значений**, вместо четырёх строк
можно записать одно выражение:

```
<значение, если истина> if <условие> else <значение, если ложь>
```

````python
x = 100

answer = "Чётное" if x % 2 == 0 else "Нечётное"
answer
````

Два случая, когда так делать не стоит: длинные выражения, которые перестают
читаться, и разные типы результата в ветвях — такое значение неудобно
использовать дальше.

````python
x = -1

# ветви возвращают разные типы: число и строку — так лучше не писать
answer = x % 2 if x > 0 else "значение отрицательное"
answer
````

## Циклы

### while

Тело выполняется, пока условие истинно. Проверка происходит перед каждой
итерацией.

```
while <условие>:
    <действия>
```

````python
n = 1

while n < 100:
    n *= 2

n
````

Если условие никогда не станет ложным, цикл не завершится. Следите за тем,
чтобы в теле менялось значение, от которого зависит условие.

### for

Цикл `for` перебирает элементы последовательности:

```
for <переменная> in <последовательность>:
    <действия>
```

````python
for i in range(1, 6):
    print(i)
````

`range(begin, end, step)` задаёт числа от `begin` включительно до `end`
не включительно с шагом `step`. Это ленивая последовательность: значения
вычисляются по мере обращения, а не хранятся в памяти целиком. Чтобы увидеть
их все сразу, `range` приводят к списку.

````python
list(range(0, 10, 2))
````

Перебирать можно любой итерируемый объект, не только `range`:

````python
for symbol in "abc":
    print(symbol)
````

````python
words = ["Hello", "World"]

for word in words:
    print(word)
````

**Не изменяйте переменную цикла внутри тела.** На следующей итерации она всё
равно получит очередное значение последовательности, поэтому такое присваивание
только запутывает читателя.

**Не изменяйте последовательность, по которой идёт перебор** — добавление и
удаление элементов во время итерации приводит к пропускам.

### Вложенные циклы

Тело цикла может содержать другой цикл:

````python
words = ["Hello", "World"]

for word in words:
    for symbol in word:
        print(symbol, end=",")
    print()
````

Если нужен и элемент, и его порядковый номер, используйте `enumerate` — он
возвращает пары «индекс, элемент». Начальный индекс задаётся аргументом
`start`.

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits, start=1):
    print(f"{index}: {fruit}")
````

`zip` объединяет несколько последовательностей поэлементно. Перебор
заканчивается вместе с самой короткой из них — остаток длинной игнорируется.

````python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
````

````python
names = ["Alice", "Bob", "Charlie"]
ages = [25]

# длина результата определяется списком ages
for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
````

### break, continue, else

* `break` — немедленно завершить цикл;
* `continue` — перейти к следующей итерации;
* `else` — блок после цикла, который выполняется, **только если цикл
  завершился без `break`**.

````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

````python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

````python
numbers = [1, 2, 3]

for number in numbers:
    if number == 6:
        print("Найдено")
        break
else:
    print("Не найдено")
````

Конструкция `for ... else` избавляет от отдельной переменной-флага «нашли или
нет». То же самое работает и с `while`.

````python
value = 257

while value > 1:
    value = value // 2 if value % 2 == 0 else value * 3 + 1
    if value == 13:
        print("Найдено 13")
        break
else:
    print("13 не встретилось")
````

### Списковые включения

Часто встречается цикл, единственная задача которого — построить список. Для
этого есть более короткая запись:

```
[<выражение> for <переменная> in <последовательность>]
```

````python
squares = []

for x in range(10):
    squares.append(x**2)

squares
````

````python
[x**2 for x in range(10)]
````

В конце можно добавить условие отбора — тогда в список попадут только
подходящие элементы:

````python
[x**2 for x in range(10) if x % 2 == 0]
````

Допустимо и несколько циклов подряд; порядок такой же, как у вложенных циклов:

````python
[(x, y) for x in range(3) for y in range(2)]
````

### any и all

Часто от последовательности нужен один ответ: подходит ли **хотя бы один**
элемент или подходят ли **все**. Для этого есть встроенные функции: `any`
истинна, если истинен хотя бы один элемент, `all` — если все.


````python
numbers = [4, 7, 10]

any([x % 2 == 0 for x in numbers]), all([x > 0 for x in numbers])
````

Обе принимают одну последовательность, поэтому условие для каждого элемента
удобно посчитать включением из предыдущего раздела.

Без `any` ту же задачу решает цикл с переменной-флагом: завести её до цикла,
выставить внутри, проверить после. Запись длиннее, а флаг легко забыть
сбросить при повторном использовании.


````python
found = False
for x in numbers:
    if x % 2 == 0:
        found = True
        break

found
````

Сами функции доходят по последовательности ровно до того места, где ответ
становится ясен: `any` останавливается на первом истинном элементе, `all` — на
первом ложном.


## match

Оператор `match` сравнивает значение с набором образцов и выполняет первую
подошедшую ветвь. Образец `_` подходит к чему угодно и играет роль `else`.

````python
language = "russian"

match language:
    case "russian":
        print("Привет")
    case "english":
        print("Hello")
    case _:
        print("Unknown language")
````

В отличие от цепочки `elif`, образцом может быть не только значение, но и
структура. Запись `[first, *rest]` означает «последовательность, первый элемент
которой равен `first`»; `*rest` собирает остаток.

````python
sequences = [[1, 2, 3], [1], [], [7, 8]]

for item in sequences:
    match item:
        case []:
            print(item, "пустая последовательность")
        case [1, *rest]:
            print(item, "начинается с единицы, дальше", rest)
        case [*_]:
            print(item, "прочая последовательность")
````

Образцы можно перечислять через `|` — подойдёт любой из них:

````python
values = [1, 5, 42]

for value in values:
    match value:
        case 1 | 2 | 3:
            print(value, "маленькое число")
        case _:
            print(value, "остальное")
````

## Работа с файлами

Данные не всегда вводятся с клавиатуры — чаще они лежат в файлах. Файл
открывается функцией `open(file, mode, encoding)`:

* `file` — путь к файлу;
* `mode` — режим работы;
* `encoding` — кодировка текста.

Режимы:

| Режим | Значение |
|---|---|
| `r` | чтение; если файла нет — ошибка |
| `w` | запись; файл создаётся, существующее содержимое стирается |
| `a` | запись в конец, не затирая прежнее |
| `r+` | чтение и запись одновременно |

Кодировку указывайте явно. По умолчанию берётся системная, и на Windows это
обычно `cp1251`, а не `utf-8` — из-за чего файлы с кириллицей читаются как
набор посторонних символов.

### Конструкция with

Открытый файл нужно закрывать: пока он открыт, часть записанных данных может
оставаться в буфере и не попасть на диск. Конструкция `with` закрывает файл
сама — в том числе если внутри блока произошла ошибка.

```
with open(...) as <переменная>:
    <работа с файлом>
```

Это предпочтительный способ. Устройство `with` разбирается в лекции про
контекстные менеджеры, пока достаточно правила: работа с файлом — внутри `with`.

````python
with open("example.txt", "w", encoding="utf-8") as file:
    file.write("Привет, мир!\n")
    file.write("Это пример работы с файлами.\n")
    file.write("Строка 3\nСтрока 4\n")

print("файл записан")
````

### Чтение целиком

Метод `read` возвращает всё содержимое одной строкой:

````python
with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

### Построчное чтение

Файл может не поместиться в память целиком, поэтому чаще его обрабатывают
построчно. Объект файла итерируемый — по нему можно идти циклом `for`, читая по
одной строке за раз.

Каждая строка приходит вместе с завершающим переводом строки; его убирает
`strip`.

````python
with open("example.txt", "r", encoding="utf-8") as file:
    for number, line in enumerate(file, start=1):
        print(f"Строка {number}: {line.strip()}")
````

Метод `readlines` читает все строки сразу в список. Он удобен для небольших
файлов, но для больших теряет преимущество построчного чтения.

````python
with open("example.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

lines
````

### Изменение содержимого

Отдельной операции «заменить часть файла» нет: файл читают целиком, меняют
строку в памяти и записывают обратно.

````python
with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()

content = content.replace("Привет", "Здравствуй")

with open("example.txt", "w", encoding="utf-8") as file:
    file.write(content)

with open("example.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

### Что делает with на самом деле

Тот же код без `with` требует закрывать файл вручную. Если этого не сделать,
данные могут остаться в буфере и следующее чтение вернёт пустоту.

````python
file = open("buffered.txt", "w", encoding="utf-8")
file.write("строка\n")

# файл ещё не закрыт — содержимое может быть не записано на диск
with open("buffered.txt", "r", encoding="utf-8") as check:
    print("до закрытия:", repr(check.read()))

file.close()

with open("buffered.txt", "r", encoding="utf-8") as check:
    print("после закрытия:", repr(check.read()))
````

Именно поэтому `with` предпочтителен: он вызывает закрытие сам, и забыть об
этом невозможно.

````python
import os

os.remove("example.txt")
os.remove("buffered.txt")
````
